```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Pada sel In [1], kode berfungsi untuk memuat pustaka-pustaka dasar yang dibutuhkan dalam pengolahan data dan algoritma pembelajaran mesin. Pustaka numpy diimpor untuk menangani operasi matriks dan array numerik, sedangkan pandas diimpor untuk mengolah data dalam bentuk tabel atau DataFrame. Selain itu, fungsi load_iris diimpor dari modul sklearn.datasets untuk mengambil dataset bawaan Bunga Iris yang siap digunakan. Pada baris terakhir sel ini, fungsi load_iris() dipanggil dan disimpan dalam variabel iris

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Pada sel In [2], data Iris yang telah dimuat diubah menjadi bentuk DataFrame Pandas dengan menentukan nama-nama kolom fiturnya menggunakan columns=iris.feature_names. Selanjutnya, dibuat kolom baru bernama Species yang memuat nama spesies asli bunga dengan memetakan indeks target ke nama kelasnya (iris.target_names[iris.target]). Setelah itu, variabel $X$ dipisahkan sebagai atribut/fitur dengan menghapus kolom Species, sedangkan variabel $y$ diambil khusus dari kolom Species sebagai label acuan. Luaran dari sel ini menampilkan isi variabel $X$ yang terdiri dari 150 baris data dan 4 kolom numerik, yaitu sepal length, sepal width, petal length, dan petal width

```python
y
```

Pada sel In [3], perintah y dijalankan untuk memeriksa isi dari variabel label target yang telah dipisahkan sebelumnya. Luaran Out [3] menampilkan daftar data target yang berupa teks/kategori nama spesies bunga, dimulai dari indeks 0 hingga 4 yang berisi spesies setosa, hingga indeks 145 hingga 149 yang berisi spesies virginica. Total data target ini berjumlah 150 baris dengan tipe data berupa objek atau teks.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Pada sel In [4], modul mengimpor kelas LabelEncoder dari sklearn.preprocessing untuk mengubah nilai label target yang berbentuk teks menjadi bentuk angka (numerik). Objek lab diinisialisasi, kemudian fungsi fit_transform(y) dijalankan untuk mengubah teks label target $y$ menjadi representasi angka secara otomatis. Luaran Out [4] memperlihatkan hasil konversi berupa array numerik, di mana angka 0 mewakili spesies setosa, angka 1 mewakili spesies versicolor, dan angka 2 mewakili spesies virginica

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Pada sel In [5], modul mengimpor fungsi train_test_split dari sklearn.model_selection untuk membagi dataset menjadi data latih (training) dan data uji (testing). Pembagian dilakukan untuk variabel fitur ($X$) dan variabel target ($y$) secara terpisah dengan proporsi data uji sebesar 20% (test_size=0.2), sehingga 80% sisa data digunakan untuk pelatihan model. Parameter random_state=0 ditambahkan agar pembagian data menghasilkan acakan yang konsisten dan dapat direplikasi setiap kali program dijalankan.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

ada sel In [6], modul memilih algoritma Gaussian Naïve Bayes karena seluruh fitur dalam dataset Iris bersifat numerik kontinu. Kelas GaussianNB diimpor dari sklearn.naive_bayes dan diinisialisasi ke dalam variabel model. Setelah itu, metode fit(X_train, y_train) dipanggil untuk melatih model Naïve Bayes menggunakan data latih yang telah dipisahkan sebelumnya, sehingga menghasilkan model terlatih bernama NB_model

```python
result = NB_model.predict(X_test)
result
```

Pada sel In [7], model yang sudah dilatih (NB_model) digunakan untuk memprediksi label kelas dari data uji X_test melalui fungsi predict(). Hasil prediksinya disimpan ke dalam variabel result. Luaran Out [7] menampilkan array angka berisi 30 nilai prediksi (berupa angka 0, 1, atau 2) yang merepresentasikan kelas hasil tebakan model untuk masing-masing sampel data uji.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Pada sel In [8], modul mengimpor fungsi accuracy_score dari sklearn.metrics untuk mengukur tingkat keakuratan dari model yang dibuat. Fungsi ini membandingkan label sebenarnya dari data uji (y_test) dengan label hasil prediksi model (result), kemudian dikalikan 100 untuk mendapatkan nilai persentase. Luaran teks menampilkan bahwa nilai akurasi pengujian model pada data uji mencapai 96,67%.
