1.Simple Linear Regression: prediksi score (y) berdasarkan waktu belajar (X) dari
dataset https://www.kaggle.com/datasets/snmahsa/student-score-suitable-for
linear-regression

```python
import pandas as pd
df = pd.read_csv('rounded_hours_student_scores.csv') 
print(df.head())
```

```python
#Menentukan predictor (X) dan target (y)
X = df.iloc[:, :-1].values  # Hours (variabel independen)
y = df.iloc[:, 1].values  # Scores (variabel dependen)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Hours')
plt.ylabel('Scores')
plt.title('Hours vs Score')
plt.show()
```

```python
#bagi dataset jadi train dan test 80:20
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=0
)
```

```python
# Fitting Simple Linear Regression to the Training set
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# memprediksi hasil pada data uji (Test set) menggunakan model regresi yang tela
y_pred = regressor.predict(X_test)
y_pred
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
score
```

```python
# Visualizing the Training set results
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
# 8. Visualisasi Hasil Training Set
plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Hours vs Scores (Training set)')
plt.xlabel('Hours')
plt.ylabel('Scores')
plt.show()
```

2.Multiple Linear Regression: prediksi sales (y) berdasarkan data marketing (x) di
youtube, facebook, newspaper, dari dataset
https://www.kaggle.com/datasets/fayejavad/marketing-linear-multiple-regression

```python
df2 = pd.read_csv('Marketing_Data.csv')
print(df2.head())
```

```python
X2 = df2.iloc[:, :-1]
y2 = df2.iloc[:, 2]
```

```python
# Bagi dataset menjadi train dan test 80:20
X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X2, y2, test_size=0.2, random_state=42
)
```

```python
regressor = LinearRegression()
regressor.fit(X_train2, y_train2)
```

```python
# Predicting the Test set results
y_pred2 = regressor.predict(X_test2)
print(y_pred2)
```

```python
# 1. Inisialisasi model regresi
regressor2 = LinearRegression()

# 2. Latih model dengan data training
regressor2.fit(X_train2, y_train2)

# 3. Baru ambil nilai intercept dan koefisiennya
a_ = regressor2.intercept_
b1 = regressor2.coef_[0]
b2 = regressor2.coef_[1]
b3 = regressor2.coef_[2]

print('Estimated model intercept, a:', a_)
print('Estimated model slope, b1 (YouTube):', b1)
print('Estimated model slope, b2 (Facebook):', b2)
print('Estimated model slope, b3 (Newspaper):', b3)
print(f'Persamaan Regresi: y_hat = {a_:.2f} + {b1:.2f}x1 + {b2:.2f}x2 + {b3:.2f}x3')
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model
score2=r2_score(y_test2,y_pred2)
score2
```
