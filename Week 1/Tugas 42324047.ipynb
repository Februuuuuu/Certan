```python
1. Buatlah sebuah Matriks berukuran 5x5 dengan elemen random. Kemudian 
elemen yang dihasilkan perlihatkan nilai maksimum, nilai minumum dan nilai rata- rata dari suatu 
elemen matriks yang di konstruksi
```

```python
import numpy as np

print("Tugas 1: Matriks 5x5 dengan Bilangan Bulat Random")
matriks = np.random.randint(0, 50, size=(5, 5))  
print("Matriksnya:")
print(matriks)
print("Nilai maksimum:", matriks.max())
print("Nilai minimum:", matriks.min())
print("Rata-rata:", matriks.mean())
```

```python
2. Buatlah grafik 2 dimensi dari persamaan berikut 
x6 + 3x4y2 - 3x4 + 3x2y4 - x2y3 - 6x2y2 +3x2 + y6 - 3y4 +3y2 =1
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("Tugas 2: Grafik 2 Dimensi")

x = np.linspace(-2, 2, 1000)
y = np.linspace(-2, 2, 1000)

X, Y = np.meshgrid(x, y)

Z = (X**6 + 3*X**4*Y**2 - 3*X**4
     + 3*X**2*Y**4 - X**2*Y**3
     - 6*X**2*Y**2 + 3*X**2
     + Y**6 - 3*Y**4 + 3*Y**2)

plt.contour(X, Y, Z, levels=[1])

plt.xlabel("x")
plt.ylabel("y")
plt.title("Grafik Persamaan")
plt.grid()
```

```python
3. Buatlah grafik 3 dimensi dari persamaan berikut
(5y/4 -[x])2 + x2
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("Tugas 3: Grafik 3 Dimensi")

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = (5*Y/4 - np.abs(X))**2 + X**2

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis')

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("Grafik 3 Dimensi")

plt.grid()
```
